```python
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import matplotlib.pyplot as plt
```

```python
root_path = 'data/OpenLane-V2_processed'
```

```python
from openlanev2.centerline.dataset import Collection, Frame

collection = Collection(root_path, root_path, 'data_dict_sample')
frame = collection.get_frame_via_identifier(('train', '00492', '315970276749927222'))
```

```python
camera = frame.get_camera_list()[0]

image = frame.get_rgb_image(camera)

plt.figure(figsize=[16, 9])
plt.imshow(image)
plt.show()
```

```python
from openlanev2.centerline.visualization import draw_annotation_bev, draw_annotation_pv, assign_attribute, assign_topology
annotations = frame.get_annotations()
annotations = assign_attribute(annotations)
annotations = assign_topology(annotations)
```

```python
image_bev = draw_annotation_bev(
    annotations, 
    with_attribute=True,
)

plt.figure(figsize=[16, 9])
plt.imshow(image_bev)
plt.show()
```

```python
from openlanev2.centerline.visualization import plot_lidar_bev_map
lidar_points, intensity = frame.get_lidar_points() 
```

```python
image_lidar_bev = plot_lidar_bev_map(lidar_points)

plt.figure(figsize=[16, 9])
plt.imshow(image_lidar_bev)
plt.show()
```
